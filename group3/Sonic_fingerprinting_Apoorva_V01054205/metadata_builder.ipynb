```python
import pandas as pd
from typing import Optional

def load_genres_dataframe(file_location: str) -> Optional[pd.DataFrame]:
    """
    Reads a tab-separated CSV file from the given location into a pandas DataFrame.

    Args:
        file_location (str): The absolute path to the tab-separated CSV file.

    Returns:
        Optional[pd.DataFrame]: A DataFrame containing the data from the CSV file if successful,
                               otherwise None if an error occurs during file reading.
    """
    try:
        df = pd.read_csv(file_location, sep='\t')
        return df
    except FileNotFoundError:
        print(f"Error: The file '{file_location}' was not found.")
        return None
    except pd.errors.EmptyDataError:
        print(f"Error: The file '{file_location}' is empty.")
        return None
    except Exception as e:
        print(f"An unexpected error occurred while reading the file: {e}")
        return None
```

```python
# Merge the DataFrames sequentially on the 'id' column using a full outer join
merged_df = pd.merge(id_genres_df, id_lang_df, on='id', how='outer')
merged_df = pd.merge(merged_df, id_information_df, on='id', how='outer')
merged_df = pd.merge(merged_df, id_metadata_df, on='id', how='outer')
merged_df = pd.merge(merged_df, id_tags_df, on='id', how='outer')

# Display the first 5 rows of the merged DataFrame
print("Merged DataFrame (id_genres_df, id_lang_df, id_information_df, id_metadata_df and id_tags_df) based on 'id' column using outer join:")
display(merged_df.head())
```

```python
# Capitalize the values in the 'genres' column
merged_df['genres'] = merged_df['genres'].str.upper()

# Capitalize the values in the 'lang' column
merged_df['lang'] = merged_df['lang'].str.upper()

# Display the first 5 rows of the DataFrame after capitalization
print("Merged DataFrame after capitalizing 'genres' and 'lang' columns:")
display(merged_df.head())
```

```python
# Define your list of master genres (ensure they are uppercase for consistent matching)
master_genres = [
    'CLASSICAL',
    'ROCK',
    'POP',
    'ELECTRONIC',
    'HIP HOP',
    'RAP',
    'JAZZ',
    'BLUES',
    'R&B',
    'SOUL',
    'FOLK',
    'COUNTRY',
    'METAL',
    'GOSPEL',
    'WORLD',
    'REGGAE',
    'LATIN',
    'PUNK',
    'FUNK',
    'DISCO',
    'INDIE',
    'ALTERNATIVE',
    'AMBIENT',
    'SOUNDTRACK',
    'EXPERIMENTAL',
    'SINGER-SONGWRITER',
    'POST-HARDCORE',
    'SHOEGAZE',
    'LO-FI',
    'EUROVISION',
    'EMO',
    'HOUSE',
    'DOWNTEMPO',
    'HARDCORE'
]

def get_primary_genre(genre_string: str, master_list: list) -> str:
    """
    Extracts a primary genre from a comma-separated genre string based on a master list.
    Prioritizes genres based on their order in the master_list.
    Uses substring matching to find master genres within individual genres.

    Args:
        genre_string (str): A comma-separated string of genres.
        master_list (list): A list of primary genres to match against.

    Returns:
        str: The first matching primary genre from the master_list, or 'OTHER' if no match.
    """
    if pd.isna(genre_string):
        return 'OTHER' # Assign 'OTHER' if genre_string is NaN

    # Split the genre string into individual genres and convert to uppercase for matching
    individual_genres = [g.strip().upper() for g in genre_string.split(',')]

    # Iterate through the master list to find the first match using substring logic
    for master_g in master_list:
        for individual_g in individual_genres:
            if master_g in individual_g: # Check if master_g is a substring of individual_g
                return master_g
    return 'OTHER' # Return 'OTHER' if no master genre is found

# Apply the function to create the new 'primary_genre' column
print("Extracting primary genres with substring matching...")
merged_df['primary_genre'] = merged_df['genres'].apply(lambda x: get_primary_genre(x, master_genres))

# Display the count of each primary genre to see the distribution
print("\nDistribution of Primary Genres (with 'OTHER' category and substring matching):")
display(merged_df['primary_genre'].value_counts())

# Display the first few rows with the new column
print("\nMerged DataFrame with 'primary_genre' column:")
display(merged_df[['id', 'genres', 'primary_genre']].head())

# Show some examples of original genres that were categorized as 'OTHER'
print("\nExamples of original genres still categorized as 'OTHER':")
display(merged_df[merged_df['primary_genre'] == 'OTHER']['genres'].value_counts().head(10))
```

```python
merged_df=merged_df[merged_df['primary_genre'] != 'OTHER']
display(merged_df.head())
```

```python
# Randomly select 1000 songs of each primary genre
# Using .sample(n=1000, random_state=42) after groupby ensures random selection and reproducibility.
metadata_df = merged_df.groupby('primary_genre', group_keys=False).apply(lambda x: x.sample(n=1000, random_state=42) if len(x) >= 1000 else x)

# Display the count of each primary genre in the new metadata_df to verify
print("Distribution of primary genres in the new 'metadata_df' (randomly sampled):")
display(metadata_df['primary_genre'].value_counts())

# Display the first few rows of the metadata_df
print("\nFirst 5 rows of 'metadata_df':")
display(metadata_df.head())
```

```python
# 1. Get the top 10 primary genres based on their counts in the original merged_df
top_10_genres = merged_df['primary_genre'].value_counts().nlargest(10).index.tolist()
print(f"Top 10 primary genres: {top_10_genres}")

# 2. Filter merged_df to include only these top 10 genres
filtered_merged_df = merged_df[merged_df['primary_genre'].isin(top_10_genres)]

# 3. Randomly select 1000 songs of each of these top 10 primary genres
# Ensure that the group has at least 1000 entries before sampling, otherwise take all available.
metadata_df = filtered_merged_df.groupby('primary_genre', group_keys=False).apply(lambda x: x.sample(n=1000, random_state=42) if len(x) >= 1000 else x)

# Display the count of each primary genre in the new metadata_df to verify
print("\nDistribution of top 10 primary genres in 'metadata_df' (randomly sampled with 1000 songs each):")
display(metadata_df['primary_genre'].value_counts())

# Display the first few rows of the metadata_df
print("\nFirst 5 rows of 'metadata_df':")
display(metadata_df.head())
```

```python
output_metadata_file_path = '/content/drive/MyDrive/music_maven_project/metadata_df.csv'
metadata_df.to_csv(output_metadata_file_path, index=False, sep='\t')
print(f"Metadata DataFrame saved successfully to '{output_metadata_file_path}'")
```
