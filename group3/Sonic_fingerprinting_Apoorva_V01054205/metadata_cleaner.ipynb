```python
result_id_genre_df = id_genre_df.groupby('primary_genre')['id_mp3'].apply(list).reset_index()
result_id_genre_df.columns = ['primary_genre', 'id_list']
result_id_genre_df.head()
```

```python
import shutil
import os

source_folder = r'C:\SID\Apoorva\Project\Music Maven Project\Output_audio_files\music4all\audios'
destination_folder = r'C:\SID\Apoorva\Project\Music Maven Project\genre_segregated_mp3_files'

for _, row in result_id_genre_df.iterrows():
    genre = row['primary_genre']
    id_list = row['id_list']          # <-- your id column is named 'id_list'
    
    genre_folder = os.path.join(destination_folder, genre)
    os.makedirs(genre_folder, exist_ok=True)
    
    for mp3_file in id_list:
        src_path = os.path.join(source_folder, mp3_file)
        dst_path = os.path.join(genre_folder, mp3_file)
        
        if os.path.exists(src_path):
            shutil.copy2(src_path, dst_path)
            print(f"Copied: {mp3_file} → {genre}/")
        else:
            print(f"Not found: {mp3_file}")

print("Done!")
```

<h1>Using Librosa library to extract features of each audio file corresponding to each genre

```python
import librosa
import numpy as np
import pandas as pd
import os
from concurrent.futures import ThreadPoolExecutor, as_completed
import warnings
warnings.filterwarnings('ignore')

def to_scalar(val):
    return float(np.squeeze(val).flat[0])

def extract_features(file_path):
    try:
        y, sr = librosa.load(file_path, sr=22050, mono=True, duration=30)
        features = {}
        features['id'] = os.path.basename(file_path)

        # ─── PRECOMPUTE SHARED COMPONENTS ─────────────────
        S        = np.abs(librosa.stft(y, n_fft=1024, hop_length=256))
        harmonic = librosa.effects.harmonic(y)

        # ─── TIMBRE ───────────────────────────────────────
        mfccs     = librosa.feature.mfcc(S=librosa.amplitude_to_db(S), sr=sr, n_mfcc=13)
        mfcc_mean = np.mean(mfccs, axis=1)
        mfcc_var  = np.var(mfccs, axis=1)
        for i in range(13):
            features[f'mfcc_{i+1}_mean'] = float(mfcc_mean[i])
            features[f'mfcc_{i+1}_var']  = float(mfcc_var[i])

        spec_centroid  = librosa.feature.spectral_centroid(S=S, sr=sr)
        spec_bandwidth = librosa.feature.spectral_bandwidth(S=S, sr=sr)
        spec_rolloff   = librosa.feature.spectral_rolloff(S=S, sr=sr)
        spec_contrast  = librosa.feature.spectral_contrast(S=S, sr=sr)

        features['spectral_centroid_mean']  = float(np.mean(spec_centroid))
        features['spectral_centroid_var']   = float(np.var(spec_centroid))
        features['spectral_bandwidth_mean'] = float(np.mean(spec_bandwidth))
        features['spectral_bandwidth_var']  = float(np.var(spec_bandwidth))
        features['spectral_rolloff_mean']   = float(np.mean(spec_rolloff))
        features['spectral_rolloff_var']    = float(np.var(spec_rolloff))

        contrast_mean = np.mean(spec_contrast, axis=1)
        contrast_var  = np.var(spec_contrast, axis=1)
        for i in range(spec_contrast.shape[0]):
            features[f'spectral_contrast_band{i+1}_mean'] = float(contrast_mean[i])
            features[f'spectral_contrast_band{i+1}_var']  = float(contrast_var[i])

        # ✅ both use matching frame_length=1024
        zcr = librosa.feature.zero_crossing_rate(y, frame_length=1024, hop_length=256)
        rms = librosa.feature.rms(y=y, frame_length=1024, hop_length=256)
        features['zcr_mean'] = float(np.mean(zcr))
        features['zcr_var']  = float(np.var(zcr))
        features['rms_mean'] = float(np.mean(rms))
        features['rms_var']  = float(np.var(rms))

        # ─── RHYTHM ───────────────────────────────────────
        onset_env    = librosa.onset.onset_strength(S=librosa.amplitude_to_db(S), sr=sr)
        tempo, beats = librosa.beat.beat_track(onset_envelope=onset_env, sr=sr)
        features['tempo_bpm'] = to_scalar(tempo)

        if len(beats) > 1:
            beat_times = librosa.frames_to_time(beats, sr=sr)
            ibi        = np.diff(beat_times)
            hist, _    = np.histogram(ibi, bins=10)
            for i, val in enumerate(hist):
                features[f'beat_histogram_bin{i+1}'] = int(val)
            features['ibi_mean'] = float(np.mean(ibi))
            features['ibi_var']  = float(np.var(ibi))
            features['ibi_min']  = float(np.min(ibi))
            features['ibi_max']  = float(np.max(ibi))
        else:
            for i in range(10):
                features[f'beat_histogram_bin{i+1}'] = 0
            features.update({'ibi_mean': 0.0, 'ibi_var': 0.0, 'ibi_min': 0.0, 'ibi_max': 0.0})

        features['onset_strength_mean']   = float(np.mean(onset_env))
        features['onset_strength_var']    = float(np.var(onset_env))
        features['onset_strength_max']    = float(np.max(onset_env))
        features['onset_strength_median'] = float(np.median(onset_env))

        # ─── HARMONY ──────────────────────────────────────
        chroma      = librosa.feature.chroma_stft(S=S, sr=sr)
        chroma_mean = np.mean(chroma, axis=1)
        chroma_var  = np.var(chroma, axis=1)
        for i in range(12):
            features[f'chroma_{i+1}_mean'] = float(chroma_mean[i])
            features[f'chroma_{i+1}_var']  = float(chroma_var[i])

        tonnetz      = librosa.feature.tonnetz(y=harmonic, sr=sr)
        tonnetz_mean = np.mean(tonnetz, axis=1)
        tonnetz_var  = np.var(tonnetz, axis=1)
        for i in range(tonnetz.shape[0]):
            features[f'tonnetz_{i+1}_mean'] = float(tonnetz_mean[i])
            features[f'tonnetz_{i+1}_var']  = float(tonnetz_var[i])

        # ─── ENERGY & DYNAMICS ────────────────────────────
        mel_spec    = librosa.feature.melspectrogram(S=S**2, sr=sr, n_mels=128)
        mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)
        features['mel_spec_mean']   = float(np.mean(mel_spec_db))
        features['mel_spec_var']    = float(np.var(mel_spec_db))
        features['mel_spec_max']    = float(np.max(mel_spec_db))
        features['mel_spec_min']    = float(np.min(mel_spec_db))
        features['mel_spec_median'] = float(np.median(mel_spec_db))

        return features

    except Exception as e:
        return {'id': os.path.basename(file_path), 'error': str(e)}


# ─── RUN ──────────────────────────────────────────────────
folder_path = r'C:\SID\Apoorva\Project\Music Maven Project\genre_segregated_mp3_files\ROCK' #change this to your actual path
output_path = 'audio_features_ROCK.parquet' #change this to your desired output path
num_workers = 6

audio_files = [
    os.path.join(folder_path, f)
    for f in os.listdir(folder_path)
    if f.endswith('.mp3')
]
total = len(audio_files)
print(f"🎵 Found {total} audio files | Running with {num_workers} threads\n")

all_features, failed = [], []

with ThreadPoolExecutor(max_workers=num_workers) as executor:
    futures = {executor.submit(extract_features, fp): fp for fp in audio_files}
    for idx, future in enumerate(as_completed(futures), 1):
        result = future.result()
        if 'error' in result:
            print(f"  [{idx}/{total}] ⚠️  {result['id']} → {result['error']}")
            failed.append(result['id'])
        else:
            print(f"  [{idx}/{total}] ✅  {result['id']}")
            all_features.append(result)

# ─── SAVE ─────────────────────────────────────────────────
features_df = pd.DataFrame(all_features)
features_df.to_parquet(output_path, index=False, compression='snappy')

print(f"\n✅ Extracted: {len(all_features)}/{total} files")
print(f"📊 Features per file: {len(features_df.columns)}")
if failed:
    print(f"❌ Failed ({len(failed)}): {failed}")
print(f"💾 Saved → {output_path}")
```

<h4> Storing the column names specific to each type ( Timbre, Rhythm, Harmony, Energy & Dynamics) in txt file -> feature_groups.txt

```python
Timbre = (
    [f'mfcc_{i}_mean' for i in range(1, 14)] +
    [f'mfcc_{i}_var'  for i in range(1, 14)] +
    ['spectral_centroid_mean', 'spectral_centroid_var',
     'spectral_bandwidth_mean', 'spectral_bandwidth_var',
     'spectral_rolloff_mean', 'spectral_rolloff_var'] +
    [f'spectral_contrast_band{i}_mean' for i in range(1, 8)] +
    [f'spectral_contrast_band{i}_var'  for i in range(1, 8)] +
    ['zcr_mean', 'zcr_var', 'rms_mean', 'rms_var']
)

Rhythm = (
    ['tempo_bpm'] +
    [f'beat_histogram_bin{i}' for i in range(1, 11)] +
    ['ibi_mean', 'ibi_var', 'ibi_min', 'ibi_max',
     'onset_strength_mean', 'onset_strength_var',
     'onset_strength_max', 'onset_strength_median']
)

Harmony = (
    [f'chroma_{i}_mean' for i in range(1, 13)] +
    [f'chroma_{i}_var'  for i in range(1, 13)] +
    [f'tonnetz_{i}_mean' for i in range(1, 7)] +
    [f'tonnetz_{i}_var'  for i in range(1, 7)]
)

Energy_and_Dynamics = [
    'mel_spec_mean', 'mel_spec_var',
    'mel_spec_max', 'mel_spec_min', 'mel_spec_median'
]

# ─── WRITE TO TEXT FILE ───────────────────────────────────
output_file = 'feature_groups.txt'

with open(output_file, 'w') as f:
    for name, cols in [('Timbre', Timbre), ('Rhythm', Rhythm),
                       ('Harmony', Harmony), ('Energy & Dynamics', Energy_and_Dynamics)]:
        f.write(f"{'═'*45}\n")
        f.write(f"  {name}  ({len(cols)} features)\n")
        f.write(f"{'═'*45}\n")
        quoted = ', '.join(f'"{c}"' for c in cols)
        f.write(f"{quoted}\n")
        f.write("\n")

print(f"✅ Saved to {output_file}")
```
