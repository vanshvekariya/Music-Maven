```python
import zipfile
import os

def extract_specific_files(zip_path, files_to_extract, output_dir, password):
    """
    Extract specific files from a password-protected zip file.

    Args:
        zip_path (str): Path to the zip file.
        files_to_extract (list of str): List of file paths inside the zip to extract.
        output_dir (str): Directory where extracted files will be saved.
        password (str): Password for the zip file.
    """
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # Convert password to bytes (required by zipfile)
    pwd = password.encode('utf-8')

    with zipfile.ZipFile(zip_path, 'r') as z:
        all_files = z.namelist()
        files_found = [f for f in files_to_extract if f in all_files]

        if not files_found:
            print("None of the requested files were found in the zip.")
            return

        for f in files_found:
            print(f"Extracting {f} ...")
            z.extract(f, path=output_dir, pwd=pwd)

        print("Extraction complete.")

# -----------------------------
# Hardcoded configuration
# -----------------------------
zip_file_path = r"C:\SID\Apoorva\Project\Music Maven Project\music4all.zip"
output_folder = r"C:\SID\Apoorva\Project\Music Maven Project\Output_files"
zip_password = "ifjlryvm2019"

files_to_extract = [
    "music4all/id_genres.csv",
    "music4all/id_information.csv",
    "music4all/id_lang.csv",
    "music4all/id_metadata.csv",
    "music4all/id_tags.csv"
]

extract_specific_files(zip_file_path, files_to_extract, output_folder, zip_password)
```

```python
import pandas as pd
from pathlib import Path

def inspect_csv_columns(csv_file_path: str):
    """
    Inspect and display all column names in the CSV file.
    Automatically detects the delimiter.
    """
    try:
        # Try reading with tab delimiter
        df = pd.read_csv(csv_file_path, sep='\t', on_bad_lines='skip')
        print(f"CSV file has {len(df.columns)} columns:")
        for i, col in enumerate(df.columns):
            print(f"  {i+1}. '{col}' (type: {df[col].dtype})")
        return df.columns.tolist()
    except Exception as e:
        print(f"Error inspecting CSV: {e}")
        return []


def csv_to_mp3_list(csv_file_path: str, id_column_name: str = 'id') -> list:
    """
    Read a TSV/CSV file with error handling, extract specified id column, convert to strings 
    with .mp3 extension, and return as a list.
    
    Args:
        csv_file_path (str): Path to the CSV/TSV file
        id_column_name (str): Name of the id column (default: 'id')
        
    Returns:
        list: List of id values converted to strings with .mp3 extension
    """
    try:
        # Read TSV file into dataframe with tab delimiter
        df = pd.read_csv(csv_file_path, sep='\t', on_bad_lines='skip')
        
        print(f"Successfully read CSV with {len(df)} valid rows")
        
        # Check if column exists
        if id_column_name not in df.columns:
            print(f"Error: Column '{id_column_name}' not found")
            print(f"Available columns: {df.columns.tolist()}")
            return []
        
        # Extract the id column
        id_column = df[id_column_name]
        
        # Define function to convert id to string and add .mp3 extension
        def convert_id_to_mp3(id_value):
            """Convert id to string and append .mp3 extension"""
            return str(id_value) + '.mp3'
        
        # Apply the function to each row of id column
        mp3_list = id_column.apply(convert_id_to_mp3).tolist()
        
        print(f"Successfully generated {len(mp3_list)} MP3 file names")
        
        return mp3_list
    
    except FileNotFoundError:
        print(f"Error: CSV file not found at {csv_file_path}")
        return []
    except Exception as e:
        print(f"Error: {e}")
        return []


# Example usage
if __name__ == "__main__":
    # Specify your CSV file path here
    csv_path = "C:\SID\Apoorva\Project\Music Maven Project\metadata_df.csv"  # Replace with your actual CSV file path
    
    # STEP 1: First, inspect the CSV to find the correct column name
    print("=" * 60)
    print("STEP 1: Inspecting CSV columns")
    print("=" * 60)
    columns = inspect_csv_columns(csv_path)
    
    # STEP 2: Use the correct column name
    print("\n" + "=" * 60)
    print("STEP 2: Converting to MP3 list")
    print("=" * 60)
    
    result = csv_to_mp3_list(csv_path, id_column_name='id')
    
    if result:
        print(f"\nGenerated {len(result)} MP3 file names:")
        for i, mp3_file in enumerate(result[:10]):  # Show first 10
            print(f"  {i+1}. {mp3_file}")
        if len(result) > 10:
            print(f"  ... and {len(result) - 10} more")
        
        # Optional: Save to a text file
        output_file = "mp3_list.txt"
        with open(output_file, 'w') as f:
            for mp3_file in result:
                f.write(mp3_file + '\n')
        print(f"\nMP3 list saved to {output_file}")
```
